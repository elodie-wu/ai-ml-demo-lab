## Part 1: Helping functions and data loading

```python
# Do not modify this cell, this is only for ploting the clusters
import matplotlib.pyplot as plt

def plot_clustering_result(X, labels, title=None):
    title = title or "clustering result"
    fig, ax = plt.subplots(figsize=(6, 5))
    noise = labels == -1
    if noise.any():
        ax.scatter(X[noise, 0], X[noise, 1], s=18, marker="x", c="k", alpha=0.9, label="noise (-1)")
    for k in sorted(set(labels) - {-1}):
        ax.scatter(X[labels == k, 0], X[labels == k, 1], s=18, alpha=0.9, label=f"cluster {k}")

    ax.set_xlabel("x1"); ax.set_ylabel("x2")
    ax.set_title(title)
    ax.legend(loc="best", frameon=True)
    ax.grid(True)
    plt.show()
```

```python
# loading data
import numpy as np
X = np.load("part1_data.npz")["X"]
```

## K-means experiments

```python
from sklearn.cluster import KMeans

# Number of clusters
K = 3

# Fit K-means and get labels
km = KMeans(n_clusters=K, random_state=42, n_init="auto").fit(X)
labels = km.labels_

# ploting data
plot_clustering_result(X, labels, title=f"K-means (k={K})")
```

I choose k=3 as the optimal number of clusters. 
The plot with k=3 captures the three main visible clusters. There is minor blending between the left end of central cluster and the bottom-left cluster. This because K-means has some limitations, it can work best when clusters are compact and roughly spherical, but central cluster is not spherical, leading to boundary misclassification.
Overall, K=3 captures the overall visible structure better than k=2(which merges distinct groups) and k=4/5(which over-segements clusters).

## DBSCAN experiments

```python
import numpy as np
from sklearn.cluster import DBSCAN

# Vary these
EPS = 1.5        # epsilon=neighbourhood radius 
MIN_POINTS = 20    # min_samples = min_points

# Fit DBSCAN and get labels
labels = DBSCAN(eps=EPS, min_samples=MIN_POINTS).fit_predict(X)

#labels: -1 = noise; 0,1,2,... = cluster IDs

# Example usage:
plot_clustering_result(X, labels, title=f"DBSCAN (eps={EPS}, min_points={MIN_POINTS})")
```

1. Three clusters with minimal noise: 
Answer: eps = 1.5, min_points = 20, number of cluster = 3

The parameters eps=1.5 and min_points=20 work well because they capture the three dense regions as valid clusters and mark sparse border points as noise. The eps radius matches the local density of the main clusters, and min_points=20 ensures only sufficiently dense regions form clusters.
